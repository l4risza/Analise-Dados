#Visualização de Dados

##Gráfico de Barras

```python
import matplotlib.pyplot as plt
```

```python
genero_counts = dados['GENERO'].value_counts()
genero_counts
```

```python
plt.figure()
plt.bar(height = genero_counts.values, x = genero_counts.index) #values(quantidade) e index(masculino, feminino)
plt.title("Quantidade de pessoas por gênero na área de Dados")
plt.xlabel('Gênero')
plt.ylabel('Contagem')
plt.grid(True)
plt.show
```

```python
import seaborn as sns #outra biblioteca para criar gráficos
```

```python
plt.figure()
sns.countplot(data = dados, x = 'GENERO', palette = 'pastel')
plt.title("Quantidade de pessoas por gênero na área de Dados")
plt.xlabel('Gênero')
plt.ylabel('Contagem')
plt.show
```

##Gráfico de Linhas

```python
salario_por_idade = dados.groupby('IDADE')['SALARIO'].mean() #média do sal por id
salario_por_idade
```

```python
plt.figure()
plt.plot(salario_por_idade.index, salario_por_idade.values, marker='o', linestyle='--')
plt.title("Média de salário por idade")
plt.xlabel('Idade')
plt.ylabel('Salário')
plt.grid(True)
plt.show
```

```python
import plotly.express as px #agora com outra biblioteca
```

```python
fig = px.line(salario_por_idade.reset_index(), x='IDADE', y='SALARIO', title='Média de salário por idade', markers=True)
fig.show()
```

##Gráfico Scatterplot

```python
plt.figure(figsize = (15, 5)) #com matplotlib
plt.scatter(dados['IDADE'], dados['SALARIO'], alpha=0.5)
plt.xlabel('Idade')
plt.ylabel('Salário')
plt.title('Dispersão de salário por idade')
plt.grid(True)
plt.show()
```

```python
#com o estilo do plotly
fig2 = px.scatter(dados, x='IDADE', y='SALARIO', title='Dispersão de salário por idade')
fig2.show()
```
